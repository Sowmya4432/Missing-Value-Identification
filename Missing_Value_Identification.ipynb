# Missing Value Identification

**Objective:** Identify missing values and summarize where they occur.

**Tools:** Python, Pandas, Matplotlib.

Upload the Titanic CSV file when prompted. The standard dataset contains 891 rows and 12 columns.

```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print('Dataset shape:', df.shape)
display(df.head())
df.info()
```

```python
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df) * 100).round(2)
missing_summary = pd.DataFrame({
    'Column Name': df.columns,
    'Missing Values': missing_count.values,
    'Missing Percentage': missing_percentage.values
})
missing_summary = missing_summary[missing_summary['Missing Values'] > 0].sort_values('Missing Values', ascending=False)
print('Missing Value Summary')
display(missing_summary)
```

```python
missing_summary.sort_values('Missing Values').set_index('Column Name')['Missing Values'].plot(kind='bar', figsize=(8, 5))
plt.title('Missing Values by Column')
plt.xlabel('Column Name')
plt.ylabel('Number of Missing Values')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print('SHORT FINDINGS')
print('Dataset rows:', df.shape[0])
print('Dataset columns:', df.shape[1])
print('Total missing cells:', int(df.isnull().sum().sum()))
print('Columns with missing values:')
print(missing_summary.to_string(index=False))
print('\nConclusion: Missing values should be investigated before further analysis. Do not remove values without justification.')
```

```python
missing_summary.to_csv('missing_value_summary.csv', index=False)
files.download('missing_value_summary.csv')
```
